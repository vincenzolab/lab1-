```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Load Datasets
air = pd.read_csv("airquality_data.csv", encoding='latin1', engine='python', on_bad_lines='skip')
heart = pd.read_csv("heart.csv", encoding='latin1', engine='python', on_bad_lines='skip')

# 2. Data Cleaning
air.fillna(air.mean(numeric_only=True), inplace=True)
heart.fillna(heart.mean(numeric_only=True), inplace=True)

air.drop_duplicates(inplace=True)
heart.drop_duplicates(inplace=True)

# 3. Data Integration (CONCAT)
air = air.reset_index(drop=True)
heart = heart.reset_index(drop=True)

min_len = min(len(air), len(heart))
air = air.iloc[:min_len]
heart = heart.iloc[:min_len]

df = pd.concat([air, heart], axis=1)

# 4. Data Transformation
scaler = StandardScaler()
num_cols = df.select_dtypes(include=np.number).columns
df[num_cols] = scaler.fit_transform(df[num_cols])

# 5. Error Correction
df.fillna(df.mean(numeric_only=True), inplace=True)

# 6. Convert Categorical → Numeric
df = pd.get_dummies(df, drop_first=True)

print("Dataset shape:", df.shape)

# 7. Data Model Building
target = 'target' if 'target' in df.columns else df.columns[-1]

X = df.drop(target, axis=1)
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
```

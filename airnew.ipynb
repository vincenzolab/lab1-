```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

# Load datasets
air = pd.read_csv("Group B/airquality_data.csv", encoding='latin1', engine='python', on_bad_lines='skip')
heart = pd.read_csv("Group B/heart.csv", encoding='latin1', engine='python', on_bad_lines='skip')

# a) Data Cleaning
air.fillna(air.mean(numeric_only=True), inplace=True)
heart.fillna(heart.mean(numeric_only=True), inplace=True)
print(air.head())

# b) Data Integration
data = pd.concat([air, heart], ignore_index=True)

# c) Data Transformation (keep numeric only)
data = data.select_dtypes(include='number')

# d) Error Correction (remove remaining nulls)
data = data.fillna(data.mean())

# e) Data Model Building
X = data.iloc[:, :-1]
y = (data.iloc[:, -1] > data.iloc[:, -1].mean()).astype(int)  # convert to 0/1

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

print("Accuracy:", model.score(X_test, y_test))
```

```python
import pandas as pd
```

```python
df = pd.read_csv('dataset_Facebook.csv', sep=';')
```

```python
print(df.head())
```

```python
# a) create data subset
```

```python
print(df.columns)
```

```python
subset_cols = df[['Page total likes', 'Type', 'Category']]
print(subset_cols.head())
```

```python
# b) merge data
```

```python
df2 = df[['Type', 'share']]
# Merge both
merged = pd.merge(df, df2, on='Type')
print(merged.head())
```

```python
#c. Sort Data
# Sort by likes
sorted_df = df.sort_values('like',ascending=False)
print(sorted_df.head())
```

```python
#d. Transpose Data
# Convert rows to columns
print(df.T.head())
```

```python
#e. Shape and Reshape Data
# Shape (rows, columns)
print(df.shape)

# Reshape using pivot
pivot = df.pivot_table(values='like', index='Type')
print(pivot)
```

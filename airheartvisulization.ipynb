```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
air = pd.read_csv("airquality_data.csv", encoding='latin1', engine='python', on_bad_lines='skip')
heart = pd.read_csv("heart.csv", encoding='latin1', engine='python', on_bad_lines='skip')

# Keep only numeric columns (IMPORTANT FIX)
air = air.select_dtypes(include='number')
heart = heart.select_dtypes(include='number')

# Fill missing values
air.fillna(air.mean(), inplace=True)
heart.fillna(heart.mean(), inplace=True)

# AIR DATASET

# Histogram
plt.figure()
air.iloc[:,0].hist()
plt.title("Air Histogram")
plt.show()

# Scatter
plt.figure()
plt.scatter(air.iloc[:,0], air.iloc[:,1])
plt.title("Air Scatter")
plt.show()


# HEART DATASET

# Boxplot
plt.figure()
sns.boxplot(x=heart.iloc[:,0])
plt.title("Heart Boxplot")
plt.show()

# Heatmap
plt.figure()
sns.heatmap(heart.corr())
plt.title("Heatmap")
plt.show()
```
